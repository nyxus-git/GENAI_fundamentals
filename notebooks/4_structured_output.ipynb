```python
from langchain_openai import ChatOpenAI
llm = ChatOpenAI(model="gpt-4o")
```

#### Discuss the Problem First

```python
text = "Hello, My name is Rohan Mane, " \
    "Email is rohanmane9841@gmail.com and my age is : 24"
```

```python
res = llm.invoke(f"please give me only name, email and age from this text: {text}")
```

```python
print(type(res.content))
```

#### Solution is: Structured Output

```python
from pydantic import BaseModel, Field
from typing import List

class ResponseStrecture(BaseModel):
    name:str = Field(description="Complete Name")
    email:str = Field(description="Email Address")
    age:int = Field(description="Age")

strectured_llm = llm.with_structured_output(ResponseStrecture)
```

```python
res = strectured_llm.invoke(f"please give me only name, email and age from this text: {text}")
```

```python
res.model_dump()
```

```python
class Movies(BaseModel):
    title:str = Field(description="Movie Title")
    year:int = Field(description="movie release year")

class AllMovies(BaseModel):
    movies:List[Movies]

movie_llm = llm.with_structured_output(AllMovies)
```

```python
res = movie_llm.invoke("Give me three sci-fi movie")
```

```python
res.model_dump()
```
