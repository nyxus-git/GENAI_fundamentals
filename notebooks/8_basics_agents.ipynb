```python
from dotenv import load_dotenv
load_dotenv()
```

```python
from langchain_core.tools import tool
```

```python
@tool
def add_number(a:int, b:int):
    """
    It will return sum of two numbers

    Args:
         a : Number One
         b : Number Twp
    """    
    return a + b


@tool
def multiply_numbers(a:int, b:int):
    """
    It will return productof two numbers

    Args:
         a : Number One
         b : Number Twp
    """    
    return a + b
```

```python
add_number.invoke({"a":12, "b":3})
```

```python
from langchain_openai import ChatOpenAI
from langchain.agents import create_agent
llm = ChatOpenAI(model="gpt-4")
```

```python
agent = create_agent(
    model = llm,
    tools=[add_number, multiply_numbers],
    system_prompt="YOur are a math teacher, and always use tool for calculations."
)
```

```python
res = agent.invoke({"messages":[{"role":"user", "content":"What is 2 + 30 * 10 ?"}]})
```
