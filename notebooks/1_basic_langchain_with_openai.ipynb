#### connect with gemini model

```python
from langchain_google_genai import ChatGoogleGenerativeAI
chat = ChatGoogleGenerativeAI(model="gemini-3.6-flash")
res = chat.invoke("Write a short poem about the beauty of nature.")
print(res.content)
```
