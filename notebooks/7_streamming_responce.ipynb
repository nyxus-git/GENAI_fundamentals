```python
from dotenv import load_dotenv
load_dotenv()
```

```python
from langchain_openai import ChatOpenAI
```

```python
llm = ChatOpenAI(model = "gpt-4", streaming = True)
```

```python
question = "Can you explain me GEN AI engineer"
```

```python
res = llm.stream(question)
for chunk in res:
    print(chunk.content, end="")
```
